#### 5.Write a Python class named Student with two attributes student_name, marks. Modify the attribute values of the said class and print the original and modified values of the said attributes.

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
stdt = Student("Muhammad Saad",50)
print("Attributes before alteration")
print(f"Name:{stdt.name}")
print(f"Marks:{stdt.marks}")
print("Attributes after alteration")
stdt.name = "Abdullah"
stdt.marks = 0
print(f"Name:{stdt.name}")
print(f"Marks:{stdt.marks}")
```
