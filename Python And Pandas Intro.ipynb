#### Problem 1
Write a function clean_name(name) that strips whitespace and returns Title Case — test it on " ali RAZA ".

```python
def clean_name(name):
    list_of_name = name.split(" ")
    result = ""
    for item in list_of_name:
        result += item.capitalize()
    return result
print(clean_name("ali Raza "))
    
```

#### Problem 3
Build a dictionary that counts how many times each letter appears in "pandas" — without using collections.Counter.

```python
word = "pandas"
count={}
for w in word:
    if w in count:
        count[w] += 1
    else:
        count[w] = 1
print(count)
```

#### Problem 4
Write a function with *args that returns both the minimum and maximum of any number of arguments as a tuple.

```python
def minMax(*numbers):
    min=numbers[0]
    max = numbers[-1]
    for number in numbers:
        if number <= min:
            min = number
        elif number >= max:
            max = number
    return (min,max)
TminMax = minMax(45, 72, 88, 91, 33, 67)
print(f"Minimum:{TminMax[0]}")
print(f"Maximum:{TminMax[1]}")
```

#### Problem 5
Use map() and filter() together (one line) to get the squares of only the even numbers from 1–20.

```python
nums = [x for x in range(1,21)]
squareOfEvens = list(map(lambda x : x**2,filter(lambda y : y%2 ==0,nums)))
print(squareOfEvens)
```

Build the same DataFrame two different ways — once from a dict of lists, once from a list of dicts — and prove they're equal with .equals().

### Problem 3
Given a DataFrame of students with a marks column, add a new column grade using np.where or .apply() with a custom function.

```python
def applyGrade(marks):
    if marks>=90 and marks<=100:
        return 'A+'
    elif marks>=80 :
        return 'A'
    elif marks>=70:
        return 'B' 
    elif marks>=60:
        return 'C'
    elif marks>=50:
        return 'D'
    else:
        return 'F'
names = ['Saad','Shahid','Hassan','Abdullah','Zamam','Hanan']
departments = ['BCS','BAI','BSE','BCS','BAI','BCS']
cgpas = [3.6,2.3,2.3,3.8,2,1.6]
DfStudents = pd.DataFrame({"Names":names,"Marks":marks,"Departments":departments,"CGPA":cgpas})
DfStudents['Grades'] = DfStudents["Marks"].apply(applyGrade)
DfStudents
```
